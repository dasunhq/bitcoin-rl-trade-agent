## Algorithm Background:
- **Policy Gradient:** Directly optimizes policy rather than value function
- **PPO:** Prevents too large policy updates using clipped objective
- **Actor-Critic:** Combines policy (actor) and value function (critic)
- **Advantage Estimation:** Uses GAE for better gradient estimates
