- **Deep Q-Network:** Uses neural network to approximate Q-values
- **Double DQN:** Reduces overestimation bias by using two networks
- **Experience Replay:** Stores transitions for stable learning

## Define DDQN Network and Agent

(Use simplified inline version for Colab)

```python
# Include full DDQN implementation here
# For brevity, showing key training loop structure

class DQNNetwork(nn.Module):
    def __init__(self, state_dim, action_dim, hidden=[256, 128, 64]):
        super().__init__()
        layers = []
        input_dim = state_dim
        for h in hidden:
            layers.extend([nn.Linear(input_dim, h), nn.ReLU(), nn.Dropout(0.2)])
            input_dim = h
        layers.append(nn.Linear(input_dim, action_dim))
        self.network = nn.Sequential(*layers)
    
    def forward(self, x):
        return self.network(x)

print("DDQN network defined")
```
